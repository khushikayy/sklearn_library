Label Encoding and Ordinal Encoding are both techniques used to convert categorical text data into numbers

Ordinal Encoding converts categorical columns into integers while assuming a specific ordered relationship between the categories.

Label Encoding converts categorical values into integers, but it treats them as purely unique identifiers without implying any order.

### using Scikitlearn

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('fuel',axis=1),df['fuel'],test_size=0.33, random_state=42)
```

```python
X_train
```

```python
y_train
```

```python
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder
```

###1. Applying Label Encoder
Recommended to use at categories under variables of output (y_train,y_test)

```python
le = LabelEncoder()
```

```python
y_train_scaled = le.fit_transform(y_train)
y_test_scaled = le.transform(y_test)
```
