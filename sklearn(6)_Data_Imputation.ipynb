A data imputer is a tool or algorithm used in statistics and machine learning to replace missing or null values in a dataset with estimated or substituted values

```python
import numpy as np
import pandas as pd
```

# Simple Imputer

```python
df
```

```python
from sklearn.impute import SimpleImputer
```

```python
simple_mean = SimpleImputer(strategy='mean')
df['A'] = simple_mean.fit_transform(df[['A']])
```
